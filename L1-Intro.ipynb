## 7. Why Classical NLP Still Matters (from the deck)

- **LLMs are black boxes that fail silently** — classical NLP gives you the tools to diagnose *why*.
- **"Does the output LOOK good?"** — evaluation is classical NLP's home turf.
- **Distributional semantics → word2vec → transformers** — understanding this progression (count-based → predictive → contextual representations) lets you place a new 2027 paper in context instead of treating each model as a magic upgrade.
- Working with LLMs is **not as simple as an API call**.

## 8. What Is NLP? Interdisciplinary Roots

NLP = the **scientific study of language**, sitting at the intersection of:

- **Linguistics** ∩ **Psychology** ∩ **Anthropology** → **Cognitive Science**
- **Computer Science** ∩ **Linguistics** → **Computational Linguistics**
- **Computational Linguistics** ∩ **AI** → feeds into **Machine Learning**, **Deep Learning**, **Computer Vision**, and **NLP**
- Approaches to NLP itself split along **Syntax-driven/Rule-based** vs. **Context-driven/Data-driven (Deep Learning)** lines, converging today in **Language Models**.

### Components of Language (Linguistics)
| Level | Studies | Example |
|---|---|---|
| **Phonology** | speech structure within a language | "that" = /th/ + /ă/ + /t/ |
| **Morphology** | smallest units of meaning (morphemes) | base words ("dog," "love") + affixes ("un-," "-ed," "-s") |
| **Syntax** | sentence structure & grammar rules | parse trees |
| **Semantics** | meaning of sentences | word/sentence meaning |
| **Pragmatics** | meaning of sentences *within context*, interaction with world knowledge | — |
| **Discourse** | how sequences of sentences (conversations, paragraphs) convey meaning | — |

**Anatomy note:** Wernicke's area → language comprehension; Broca's area → speech production, grammar, motor planning for articulation.

## 9. History of NLP (Timeline)

Weaver's Memorandum → Symbolic Approaches / Ontologies → **AI Winter in NLP** → Statistical Models → Neural Word Embedding → Seq2Seq / Attention → **Transformers (2017)** → Multi-task Learning → GPT → **LLMs (2020s)** → Small Language Models (SLM, 2024) → **Agentic AI (2026)**

### AI Winter in NLP (key takeaway)
Early machine translation reports (ALPAC-era) recommended **less** focus on full end-to-end MT and **more** focus on aiding human translators — practical evaluation methods, cataloguing translation resources, and speeding up human translation workflows.

### Modern era additions (post-2023, not in most Stanford lecture archives)
- **Small Language Models (SLM):** modest parameter counts (millions–low billions) vs. LLMs (hundreds of billions–trillions). Examples: DistilBERT, Gemma, GPT-4o mini, Granite, Llama, Ministral, Phi.
- **Agentic AI:**
  - Shift to *reasoning models* (e.g., GPT-o1) using inference-time scaling / intermediate reasoning steps
  - Sparse & multimodal "omni" models (e.g., GPT-4o)
  - The "context revolution": long-context windows vs. RAG
  - **Skills:** self-contained instruction/resource folders an agent loads on demand — turns an LLM from a "text generator" into a goal-directed system
  - Shift from chatbots to **agentic workflows**

## 10. Challenges of Natural Language Processing

**Productivity** (Lyons, 1977): *"property of the language-system which enables native speakers to construct and understand an indefinitely large number of utterances, including utterances they have never previously encountered."*

### Ambiguity — by level
| Level | Definition | Example |
|---|---|---|
| Phonological | words sound alike | Right/Write, Knows/Nose |
| Lexical | more than one meaning for a word | Bank (financial vs. riverbank) |
| Syntactic | more than one sentence-level interpretation | — |
| Semantic | more than one meaning given wider context (polysemy) | "set," "arm," "head"; "Head of New Zealand is a woman"; "Michael Jordan is a professor at Berkeley"; "your cherry coat" |
| Referential | unclear which entity a word/phrase refers to | "Alice told Jane that **her** car was parked illegally." |
| Pragmatic | ambiguity from context/conversational use, not structure | "I didn't tell John you were late." |

### Variability — by level
Phonetic (accent), Morphological, Lexical (spelling), Syntactic

### Diversity
Phonologic diversity, Graphemic diversity, Syntactic diversity (Russell S. Tomlin, 1980)

### Sparsity
A large number of language units occur with very low frequency — long-tail distribution problem.

## 11. Real-World NLP Use Cases (from the deck)

- **Named Entity Recognition**
- **Machine Translation** — from 49 hand-picked Russian sentences (1954, Georgetown-IBM experiment) to full production MT systems (2023)
- **Text Summarization** — e.g., condensing a clinical note (patient history, vitals, meds, follow-up plan) into a compact summary
- **Information Retrieval** — query representation + corpus representation → indexing → matching
- **Chatbots** — including domain-specific bots (e.g., pulmonary diagnosis screening assistant)

### Clinical NLP Use Case (deep dive)
- **Objective:** automate structuring of patient health records to human-level performance
- **Inputs:** PDFs, fax copies, document images, HL7 messages, ...
- **Outputs:** structured fields across Lab, Path, Radiology, Genetics reports
- **Challenges:**
  - Capture format (handwritten/digital, templated/free-style)
  - Quality (digital PDF vs. scanned)
  - Data format (free-text, tables, key-value pairs)
  - Document variety (radiology, pathology, lab results)
  - Technical language variation between practitioners
  - Multilingual text (English, French, Spanish, etc.)
  - Data size, variety, and label availability
